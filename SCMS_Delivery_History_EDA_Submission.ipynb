# Project Summary

This project explores the SCMS Delivery History Dataset, which records shipment and delivery line items for health-related products. The dataset contains **10,324 rows and 33 original columns**, covering countries, vendors, product groups, shipment modes, dates, quantities, prices, freight, and insurance. The goal of the analysis is to understand how products move through the supply chain, where order value and volume are concentrated, and what the data suggests about delivery performance and logistics costs.

The initial data audit found **no exact duplicate rows**. Three original columns contain missing values: Shipment Mode (360 rows), Dosage (1,736 rows), and Line Item Insurance (USD) (287 rows). Several fields that look numeric, including Weight (Kilograms) and Freight Cost (USD), are stored as text because they include labels such as “Weight Captured Separately” and “Freight Included in Commodity Cost”. These are converted to numeric helper columns for analysis while preserving the original values. Date columns also contain process labels such as “Pre-PQ Process” or “Date Not Captured”; invalid date strings are coerced to missing dates rather than interpreted as real dates.

The data is strongly concentrated in the **ARV product group**, which represents 8,550 of the 10,324 line items. Across all line items, the recorded Line Item Value totals approximately **USD 1.63 billion**. South Africa has the highest number of records among countries (1,406), followed by Nigeria (1,194) and Côte d’Ivoire (1,083). Air is the most frequently recorded shipment mode (6,113 records), followed by Truck (2,830). Shipment mode is not available for 360 records, so mode-based comparisons should be interpreted with this limitation in mind.

Delivery timing is assessed by calculating the number of days between the scheduled delivery date and the delivered-to-client date. In this dataset, 1,186 records have a positive delay, while 9,138 are on or before the scheduled date. The mean delay is affected by early deliveries and a small number of large delays, so both the distribution and the share of late records are useful measures. These figures describe the recorded line items, not necessarily unique shipments or service-level performance at a company-wide level.

The notebook uses summary statistics and visualizations to examine product mix, country-level activity, shipment modes, yearly delivery volume, vendor concentration, order values, quantities, lead/delay patterns, and numeric relationships. The final recommendations focus on improving data quality, monitoring late deliveries, reviewing high-value product flows, and evaluating freight costs by shipment mode. The results are descriptive and observational: they can guide further investigation, but they do not prove that any one factor causes delivery delays or higher costs.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Create analysis-ready helper columns while preserving original columns.
import numpy as np
import pandas as pd

clean_df = df.copy()

date_columns = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]
for col in date_columns:
    clean_df[col + ' Parsed'] = pd.to_datetime(clean_df[col], errors='coerce', dayfirst=True)

clean_df['Weight (Kilograms) Numeric'] = pd.to_numeric(
    clean_df['Weight (Kilograms)'], errors='coerce'
)
clean_df['Freight Cost (USD) Numeric'] = pd.to_numeric(
    clean_df['Freight Cost (USD)'], errors='coerce'
)
clean_df['Delivery Delay Days'] = (
    clean_df['Delivered to Client Date Parsed'] -
    clean_df['Scheduled Delivery Date Parsed']
).dt.days
clean_df['Delivery Year'] = clean_df['Delivered to Client Date Parsed'].dt.year
clean_df['Is Late'] = clean_df['Delivery Delay Days'] > 0

# Standardize whitespace in text columns without changing missing values.
text_cols = clean_df.select_dtypes(include='object').columns
for col in text_cols:
    clean_df[col] = clean_df[col].map(lambda x: x.strip() if isinstance(x, str) else x)

print('Analysis-ready shape:', clean_df.shape)
print('Valid delay observations:', clean_df['Delivery Delay Days'].notna().sum())
print('Numeric freight observations:', clean_df['Freight Cost (USD) Numeric'].notna().sum())
print('Late line items:', int(clean_df['Is Late'].sum()))
````

### What all manipulations have you done and insights you found?

The original data was retained in `df`, and a working copy named `clean_df` was created. Date fields were parsed with invalid process labels converted to `NaT`; numeric helper columns were created for weight and freight cost; delivery delay was calculated as delivered date minus scheduled date; delivery year and a late-delivery flag were added; and leading/trailing spaces in text fields were stripped. Missing values were not automatically filled with zero because a missing value does not necessarily mean that the true value is zero. This preserves the distinction between unavailable data and actual zero values.
